# Dataset inspection
Run `python scripts/inspect_dataset.py --config configs/base.yaml` from the repository root first. This notebook reads generated outputs; it does not manufacture observations.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
manifest = json.loads((root / "outputs/inventory/dataset_manifest.json").read_text())
manifest["status"], manifest["file_count"], manifest["battery_ids"]
```

```python
files = pd.DataFrame(manifest["files"])
display(files.groupby(["status", "extension"], dropna=False).size().rename("files"))
display(files[["source_file", "cell_id", "test_type", "duty_cycle_type", "status"]])
```

```python
tables = pd.DataFrame([{ "source_file": file["source_file"], **table} for file in manifest["files"] for table in file["tables"]])
display(tables[["source_file", "name", "detected_columns", "scan_complete", "number_of_observations"]])
```
